# WM-811K Exploratory Data Analysis

Class distribution, wafer map size variability, and sample maps per defect type.


```python
import sys
sys.path.append('..')

from src.data_loader import load_raw, clean, labeled_subset
from src.visualize import plot_class_distribution, plot_sample_grid
from src import config

df = load_raw(config.RAW_PKL)
df = clean(df)
print(f'Total wafer maps: {len(df):,}')
df.head()
```

```python
labeled = labeled_subset(df)
print(f'Labeled wafer maps: {len(labeled):,} ({len(labeled)/len(df):.1%} of total)')
labeled['failureType'].value_counts()
```

```python
# Class distribution is heavily skewed toward 'none' -- this is why
# macro-F1 / balanced accuracy matter more than raw accuracy for this task.
plot_class_distribution(labeled)
```

```python
# Wafer map dimensions vary quite a bit -- this is why we resize before
# feeding maps into a fixed-size CNN input.
labeled['waferMapDim'].value_counts().head(10)
```

```python
plot_sample_grid(labeled, n_per_class=4)
```
